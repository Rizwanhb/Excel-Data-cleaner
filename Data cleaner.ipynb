```python
import pandas as pd
from datetime import datetime
```

```python
#Just for decorations
print("="*50)
print("EXCEL DATA CLEANER".center(50))
print("="*50)

#Read the messy data
df = pd.read_excel('messy-data.xlsx')




#Cleaning process

#1. Strip whitespaces from all text to column
df = df.map(lambda x: " ".join(x.split()) if isinstance(x, str) else x)


#2. Lowercase all columns 
df.columns=[x.lower() for x in df.columns]


#3.Replace spaces and symbols with undersore.
df = df.replace(' ', '_', regex=True)


#4.Handling columns with same name
s = pd.Series(df.columns)
df.columns = s + s.groupby(s).cumcount().astype(str).replace('0', '')


#5.Drop empty rows and columns
df = df.dropna(axis=1, how='all').dropna(axis=0, how='all')
df

#6. Check for missing values across the dataset
print(df.isnull().sum())


#7.Remove duplicate rows
df = df.drop_duplicates()

#8.Convert numbers stored as strings into int or float
for col in df.columns:
    df[col]=pd.to_numeric(df[col], errors='ignore')


#9.Convert Date to Datetime formate
s = df.iloc[:, 0]
final_dates = pd.to_datetime(s.str[:3] + ' 2001', format='%b %Y', errors='coerce')

print(final_dates)


df
```

"As shown above, an assessment of the dataset reveals no missing values in the numerical columns, so no imputation or data dropping was necessary for this phase."

```python
df.to_excel("Cleaned_data.xlsx", index=False)
print("Saved!")
```
